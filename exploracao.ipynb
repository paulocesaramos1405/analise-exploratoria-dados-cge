```python
df_tratado = df.copy()
```

```python
df_tratado.columns = (
    df_tratado.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)
    
```

```python
df_tratado["data_abertura"] = pd.to_datetime(
    df_tratado["data_abertura"],
    errors="coerce"
)

df_tratado["data_conclusao"] = pd.to_datetime(
    df_tratado["data_conclusao"],
    errors="coerce"
)
```

```python
df_tratado.info()
```

```python
df_tratado["canal"] = (
    df_tratado["canal"]
    .str.strip()
    .str.title()
)
```

```python
df_tratado["canal"].value_counts()
```

```python
df_tratado[df_tratado["municipio"].isnull()]
```

```python
df_tratado["municipio"] = (
    df_tratado["municipio"].fillna("Não informado")
)
```
